```python
import pandas as pd

# DATA SNLIK OJK 2013-2025
# Sumber: OJK Siaran Pers SNLIK berbagai tahun

# 1. Tren indeks nasional per tahun
tren_nasional = pd.DataFrame({
    'tahun': [2013, 2016, 2019, 2022, 2025],
    'literasi': [21.84, 29.66, 38.03, 49.68, 66.46],
    'inklusi': [59.74, 67.82, 76.19, 85.10, 80.51]
})

# 2. Data 2025 per wilayah
wilayah = pd.DataFrame({
    'wilayah': ['Urban', 'Rural'],
    'literasi': [70.89, 59.60],
    'inklusi': [83.61, 75.70]
})

# 3. Data 2025 per jenis kelamin
gender = pd.DataFrame({
    'gender': ['Laki-Laki', 'Perempuan'],
    'literasi': [67.32, 65.58],
    'inklusi': [80.73, 80.28]
})

# 4. Data 2025 per sektor keuangan
sektor = pd.DataFrame({
    'sektor': ['Perbankan', 'Perasuransian', 'Pergadaian',
               'Lembaga Pembiayaan', 'Dana Pensiun',
               'Pasar Modal', 'Lembaga Keuangan Mikro'],
    'literasi': [65.50, 45.45, 54.74, 46.66, 27.79, 17.78, 9.80],
    'inklusi': [70.65, 28.50, 8.23, 12.38, 5.37, 1.34, 1.20]
})

# 5. Data 2025 per kelompok usia
usia = pd.DataFrame({
    'kelompok_usia': ['15-17 th', '18-25 th', '26-35 th', '36-50 th', '> 50 th'],
    'jumlah_responden': [585, 1268, 2021, 3429, 3442]
})

# 6. Data 2025 per pendidikan
pendidikan = pd.DataFrame({
    'pendidikan': ['Tidak Bersekolah', 'SD', 'SMP', 'SMA', 'Perguruan Tinggi'],
    'jumlah_responden': [1464, 2384, 2103, 3524, 1270]
})

print("Semua data berhasil dibuat!")
print(f"\nTren Nasional:\n{tren_nasional}")
print(f"\nPer Wilayah:\n{wilayah}")
print(f"\nPer Sektor:\n{sektor}")
```

```python
#EXPORT DATA KE CSV UNTUK POWER BI
import os

os.makedirs('../03-inklusi-keuangan/data', exist_ok=True)

tren_nasional.to_csv('../03-inklusi-keuangan/data/tren_nasional.csv', index=False)
wilayah.to_csv('../03-inklusi-keuangan/data/wilayah.csv', index=False)
gender.to_csv('../03-inklusi-keuangan/data/gender.csv', index=False)
sektor.to_csv('../03-inklusi-keuangan/data/sektor.csv', index=False)
usia.to_csv('../03-inklusi-keuangan/data/usia.csv', index=False)
pendidikan.to_csv('../03-inklusi-keuangan/data/pendidikan.csv', index=False)

print("Semua file CSV berhasil diexport!")
print("\nFile yang tersimpan di folder data/:")
for f in os.listdir('../data'):
    print(f"  - {f}")
```
